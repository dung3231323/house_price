```python
import pandas as pd
import pyodbc
from sqlalchemy import create_engine
```

```python
# create connection to SQL server
server = 'DUNG'
database = 'learning'
trusted_connection='yes'
connection_string = f"mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
engine = create_engine(connection_string)
```

```python
file_name='https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/FinalModule_Coursera/data/kc_house_data_NaN.csv'
df = pd.read_csv(file_name)
df.to_sql('sales', con=engine, schema='staging', if_exists='replace', index=False)
```

```python
df.head(5)
```

```python
df.drop(['Unnamed: 0', 'id'], axis = 1, inplace = True)
df.describe()
```

```python
df['date'] = pd.to_datetime(df['date'], format='%Y%m%dT%H%M%S')
df.date
```

```python
print("Number of null value in df.date:",df.date.isnull().sum())
print("Number of null value in df.yr_built:", df.yr_built.isnull().sum())
print("Number of null value in df.yr_renovated:", df.yr_renovated.isnull().sum())
```

```python
columns = df.columns.drop(['date', 'yr_built', 'yr_renovated'])
for column in columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mean())

df.isnull().sum()
```

```python
df.to_sql('sales', con=engine, schema='house_star', if_exists='replace', index=False)
```

```python
df.to_csv("sales.csv")
```

```python
conn = pyodbc.connect(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=DUNG;"        
    "DATABASE=learning;"          
    "Trusted_Connection=yes;"   
)

cursor = conn.cursor()

with open("./house_star_schema.sql", "r", encoding="utf-8") as f:
    sql_script = f.read()

commands = sql_script.split('GO')  # split by GO keyword

for command in commands:
    cmd = command.strip()
    if cmd:
        try:
            cursor.execute(cmd)
            conn.commit()
        except Exception as e:
            print(f"Lỗi khi chạy lệnh:\n{cmd}\n==> {e}")

cursor.close()
conn.close()
```
